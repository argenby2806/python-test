```python
##Print Formatting with Strings
#f strings
s = "hello"
print(f"Hello, {s}!") #Hello, hello!
#f.format()
#mặc định 
print(" {} {} {} ".format("nem", "chua", "Thanh Hóa")) # nem chua Thanh Hóa
#định dạng theo vị trí
print(" {2} {1} {0} ".format("nem", "chua", "Thanh Hóa")) # Thanh Hóa chua nem
#định dạng theo tên 
print(" {food} {dish} {place} ".format(food="nem", dish="chua", place="Thanh Hóa")) # nem chua Thanh Hóa
```
