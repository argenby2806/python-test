```python
with open('test_file.txt', mode='w') as f:
    f.write('Dong 1: Chieu nay di hoc Python.\n')
    f.write('Dong 2: Khai niem mode rat quan trong.\n')

print("Đã tạo file test_file.txt thành công!")
    
```

```python
with open('test_file.txt', mode='r') as f:
    content = f.read()
    print("Nội dung của file test_file.txt:")
    print(content)
```

```python
with open('test_file.txt', mode='a') as f:
    f.write('Dong 3: Day la dong duoc them vao.\n' )
with open('test_file.txt', mode='r') as f:
    print("Nội dung của file test_file.txt sau khi thêm dòng:")
    print(f.read())
```

```python
with open('test_file.txt', mode = 'w') as f:
    f.write('Toan bo noi dung cu da bi xoa sach!\n')
with open('test_file.txt', mode = 'r') as f:
    print("Nội dung của file test_file.txt sau khi xóa:")
    print(f.read())
```

```python
with open('test_file.txt',mode = 'r+') as f:
    first_line = f.readline()
    print("Dong dau tien cua file test_file.txt:")  
    f.write('Dong moi duoc them tu mode r+\n')
with open('test_file.txt', mode = 'r') as f:
    print("Nội dung của file test_file.txt sau khi thêm dòng mới:")
    print(f.read())
```
